**Question 2: What are the key assumptions of Simple Linear Regression?**

Simple Linear Regression (SLR) is based on certain assumptions that must be satisfied for the model to produce reliable and accurate results. The key assumptions are as follows:

1. Linearity

There is a linear relationship between the independent variable and the dependent variable. This means that changes in the independent variable result in proportional changes in the dependent variable.

2. Independence of Errors

The residuals (errors) are independent of each other. The error for one observation should not be related to the error of another observation.

3. Homoscedasticity

The variance of the residuals is constant across all values of the independent variable. In other words, the spread of errors should remain the same throughout the dataset.

4. Normality of Errors

The residuals should be normally distributed, especially important for hypothesis testing and confidence interval estimation.

5. No Multicollinearity

Since Simple Linear Regression involves only one independent variable, multicollinearity is not an issue. However, this assumption becomes important in multiple linear regression.

**Conclusion:**

If these assumptions are met, Simple Linear Regression provides unbiased, efficient, and meaningful estimates, making it a powerful tool for prediction and analysis.

**Question 3: Write the mathematical equation for a simple linear regression model and explain each term.**

The mathematical equation of a Simple Linear Regression (SLR) model is:

Y=β0​+β1​X+ε

**Explanation of Each Term:**

- **Y (Dependent Variable):** The outcome or response variable that we want to predict or explain.

- **X (Independent Variable):** The predictor or input variable that is used to explain changes in the dependent variable.

- **β0 (Intercept):** The value of the dependent variable Y when the independent variable X is zero. It represents the starting point of the regression line on the Y-axis.

- **β1 (Slope or Regression Coefficient):** Represents the change in Y for a one-unit increase in X. It indicates the direction and strength of the relationship between X and Y.

- **ε (Error Term):** Accounts for the variation in Y that cannot be explained by X. It represents random noise or unexplained factors affecting the dependent variable.

**In Simple Words**

The equation shows that the dependent variable Y is explained as a linear function of the independent variable X, along with some random error.

**Question 5: What is the method of least squares in linear regression?**

The method of least squares works by minimizing the sum of the squares of these residuals. Squaring the residuals ensures that both positive and negative errors contribute positively and that larger errors are penalized more heavily.

**Question 6: What is Logistic Regression? How does it differ from Linear Regression?**

Logistic Regression is a statistical and machine learning technique used for classification problems, where the dependent variable is categorical, usually binary (such as Yes/No, 0/1, True/False). It predicts the probability that an observation belongs to a particular class using a logistic (sigmoid) function.

**Logistic Regression Explained:**

Logistic Regression models the relationship between independent variables and the probability of an outcome. Instead of predicting a continuous value, it predicts a value between 0 and 1, which is then mapped to a class.





#**Difference Between Logistic Regression and Linear Regression**



| Feature            | Linear Regression        | Logistic Regression         |
| ------------------ | ------------------------ | --------------------------- |
| Type of Problem    | Regression               | Classification              |
| Dependent Variable | Continuous               | Categorical (Binary)        |
| Output Value       | Any real number          | Probability between 0 and 1 |
| Function Used      | Linear function          | Sigmoid (logistic) function |
| Error Metric       | Mean Squared Error (MSE) | Log Loss (Cross-Entropy)    |
| Typical Use        | Predicting prices, marks | Predicting yes/no outcomes  |


**Conclusion**

Linear Regression is used when the output variable is continuous, whereas Logistic Regression is used when the output variable is categorical. Logistic Regression is especially useful for probability prediction and classification tasks.

**Question 7: Name and briefly describe three common evaluation metrics for regression models.**

Three commonly used evaluation metrics for regression models are:

**1. Mean Absolute Error (MAE)**

Mean Absolute Error measures the average absolute difference between the actual values and the predicted values. It treats all errors equally and is easy to interpret because it is expressed in the same units as the target variable.

**2. Mean Squared Error (MSE)**

Mean Squared Error calculates the average of the squared differences between actual and predicted values. By squaring the errors, it gives more importance to larger errors, making it useful when large mistakes are particularly undesirable.

**3. Root Mean Squared Error (RMSE)**

Root Mean Squared Error is the square root of MSE. It represents the average prediction error in the same units as the dependent variable and is more sensitive to large errors than MAE.

**Conclusion: **

These metrics help evaluate how well a regression model performs by quantifying the difference between predicted and actual values.

**Question 8: What is the purpose of the R-squared metric in regression analysis?**

R-squared (R²) is a statistical measure used to determine how well a regression model explains the variability of the dependent variable.

**Purpose of R-squared**

The main purposes of the R-squared metric are:

- It indicates the proportion of variance in the dependent variable that is explained by the independent variable(s).
- It helps assess the goodness of fit of the regression model.
- It allows comparison between different regression models fitted on the same dataset.

R-squared values range from 0 to 1:

- R² = 0 means the model explains none of the variability.
- R² = 1 means the model explains all the variability.

**Conclusion:**

In simple terms, R-squared shows how well the regression model fits the data and how much of the outcome can be explained by the predictors.

**Question 9: Write Python code to fit a simple linear regression model using scikit-learn and print the slope and intercept.**


```python
# Import required libraries
import numpy as np
from sklearn.linear_model import LinearRegression

# Sample data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # Independent variable
y = np.array([2, 4, 6, 8, 10])               # Dependent variable

# Create and fit the model
model = LinearRegression()
model.fit(X, y)

# Print slope and intercept
print("Slope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)
```
